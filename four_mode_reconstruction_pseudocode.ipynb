# Four-mode temperature reconstruction pseudocode

This notebook-form document defines Baseline, Scaled, fixed one-anchor, and clipped fixed two-anchor reconstruction. It is implementation-neutral pseudocode, not executable software. All hours are UTC.

```text
PSEUDOCODE
FOR h = 0,...,23:
    S[h] = A1 * sin(2*pi*h/24 - phi1)
         + A2 * sin(2*pi*h/12 - phi2)

baseline[h] = daily_mean + S[h]
```

```text
PSEUDOCODE
IF fitted_DTR >= 0.5 degrees Celsius:
    beta_scaled = daily_DTR / fitted_DTR
ELSE:
    beta_scaled = 1
scaled[h] = daily_mean + beta_scaled * S[h]
```

```text
PSEUDOCODE
alpha_one = observed_temperature[12] - S[12]
one_anchor[h] = alpha_one + S[h]
primary one-anchor scores exclude hour 12
```

```text
PSEUDOCODE
denominator = S[12] - S[0]
IF absolute_value(denominator) <= 0.5 degrees Celsius:
    beta = 1
    alpha = mean(observed_temperature[0] - S[0],
                 observed_temperature[12] - S[12])
ELSE:
    beta_raw = (observed_temperature[12] - observed_temperature[0]) / denominator
    beta = clip(beta_raw, 0.1, 5.0)
    alpha = observed_temperature[12] - beta * S[12]
two_anchor[h] = alpha + beta * S[h]
primary two-anchor scores exclude hours 00 and 12
```
